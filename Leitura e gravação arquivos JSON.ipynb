#### Listar os arquivos JSON no volume


```
%python
dbutils.widgets.text("volume_dir", "")
```

```
%python
volume_dir = dbutils.widgets.get("volume_dir").strip().rstrip("/")
if not volume_dir.startswith("/Volumes/") or len(volume_dir.split("/")) < 5:
    raise ValueError("Informe /Volumes/catalogo/esquema/volume no widget volume_dir.")

events_dir = f"{volume_dir}/events"
display(dbutils.fs.ls(events_dir))
```

#### Exibindo um arquivo Json com as informações

```
%python
print(dbutils.fs.head(f"{events_dir}/file-1.json"))
```

#### Carregando 1 arquivo Json para o dataframe

```
%python
dataf = spark.read.json(f"{events_dir}/file-1.json")
dataf.printSchema()
dataf.show()
```

#### Carregando 2 arquivos Json para o dataframe

```
%python
dataf2 = spark.read.json([f"{events_dir}/file-1.json", f"{events_dir}/file-2.json"])
dataf2.show()
```

#### Carregando TODOS os arquivos Json para o dataframe

```
%python
dataf3 = spark.read.json(f"{events_dir}/*.json")
dataf3.show()
```

#### Gravar os resultados em um diretório do volume


```
%python
from uuid import uuid4

output_dir = f"{volume_dir}/saida/json_unificado_{uuid4().hex}"
dataf3.write.json(output_dir)
```

#### Criação de uma tabela para executar SQL

```
%python
spark.read.json(output_dir).createOrReplaceTempView("view_evento")
spark.sql("SELECT action FROM view_evento").show()
```
