### Carregando arquivo CSV em python

```
%python
dbutils.widgets.text("volume_dir", "")
```

```
%python
volume_dir = dbutils.widgets.get("volume_dir").strip().rstrip("/")
if not volume_dir.startswith("/Volumes/") or len(volume_dir.split("/")) < 5:
    raise ValueError("Informe /Volumes/catalogo/esquema/volume no widget volume_dir.")

clientes = spark.read.csv(f"{volume_dir}/clientes_cartao.csv", header=True, inferSchema=True, sep=";")
display(clientes)
```

### Carregando arquivo CSV em Scala

```
%scala
val volumeDir = dbutils.widgets.get("volume_dir").trim.stripSuffix("/")
val cliente = spark.read.format("csv")
  .option("header", "true")
  .option("inferSchema", "true")
  .option("delimiter", ";")
  .load(s"$volumeDir/clientes_cartao.csv")
display(cliente)
```

### Trabalhando com o arquivo carregado em Scala no SQL

```
%scala
cliente.createOrReplaceTempView("dados_cliente")
```

### Exibindo os dados no SQL

```
%sql
select * from dados_cliente
```
